# Ein- und Ausgaben

Eingabe und Ausgabe sind in Python sehr einfach. Die Eingabe erfolgt über die Funktion `input()`, die Ausgabe über `print()`.
Die Eingabe wird immer als String gespeichert. Um sie in einen anderen Datentyp zu konvertieren, verwenden Sie die Funktionen `int()`, `float()` oder `str()`.
Variablen und Ausdrücke lassen sich über F-Strings direkt in Strings einbinden. Sie sind sehr praktisch und lesbar. Beispiel: `f"Mein Name ist {name} und ich bin {age} Jahre alt."`

F-Strings lassen sich auch formatieren. Das ist besonders wichtig, wenn Sie Zahlen ausgeben wollen. Hier einige Beispiele:
```python
var = 1.23456789
print(f"{var:.2f}")  # Ausgabe: 1.23
print(f"{var:.2e}")  # Ausgabe: 1.23e+00
print(f"{var:+}")    # Ausgabe: +1.23456789
```



Bedeutung | Befehl 
-------- |--------
Eingabe auslesen und in Variable `var` speichern   | `var = input("message")` 
Ausgabe von `var`   | `print(var)` 
Formatierung von `var` als Gleitkommazahl mit 2 Stellen nach dem Komma | `f"{var:.2f}"`
Formatierung von `var` in Exponentialdarstellung mit 2 Stellen nach dem Komma | `f"{var:.2e}"`
Formatierung von `var` mit positivem und negativem Vorzeichen | `f"{var:+}"`

```python

elif andere_bedingung: # optional
    # Code-Block, der ausgeführt wird, wenn die andere Bedingung True ist
```

# Sequenzielle Datentypen

Neben Listen gibt es auch noch Tupel, Mengen und Dictionaries. Diese sind in Python sehr wichtig. Hier sind einige Beispiele:
1. Listen (list)
Veränderbare, geordnete Sammlungen von Elementen beliebiger Typen.
```python
liste = [1, 2, 3, "vier", 5.0]
liste[0:3]  # Zugriff auf die ersten drei Elemente
liste[-1]  # Zugriff auf das letzte Element
liste.append(6)  # Element hinzufügen
liste.remove(2)  # Element entfernen
```
2. Tupel (tuple)
Unveränderbare, geordnete Sammlungen von Elementen beliebiger Typen.
```python
tupel = (1, 2, 3, "vier", 5.0)
tupel[0]  # Zugriff auf das erste Element
tupel[0]= 2  # Fehler: Tupel sind unveränderbar
```
3. Mengen (set)
Ungeordnete Sammlungen von einzigartigen Elementen.
```python
menge = {1, 2, 3, "vier", 5.0}
menge.add(6)  # Element hinzufügen
menge.remove(2)  # Element entfernen
```
4. Dictionaries (dict)
Ungeordnete Sammlungen von Schlüssel-Wert-Paaren.
```python
dictionary = {"name": "Max", "alter": 25}
dictionary["name"]  # Zugriff auf den Wert mit dem Schlüssel "name"
dictionary["name"] = "Moritz"  # Wert ändern
```
5. Strings (str)
Strings sind unveränderbare Sequenzen von Zeichen.
```python
string = "Hallo, Welt!"
string[0]  # Zugriff auf das erste Zeichen
string[0] = "h"  # Fehler: Strings sind unveränderbar
```

Bedeutung | Befehl 
-------- |--------
 Subliste von Index `0` bis `10` jedes `2`-te Element  | `l[0:10:2]`
 Vorletztes Element einer Liste | `l[-2]`
 Tupel aus zwei Listen (notwendigerweise gleich lang) | `zip(l1,l2)` 
List Comprehension: erzeugen einer neuen Liste | `l2 = [x**2 for x in range(20)]` 
String an Leerzeichen aufsplitten | `"Hallo Welt".split(" ")`

# Schleifen, Funktionen und Dateien
## Schleifen
Schleifen ermöglichen es, Blöcke von Code mehrfach auszuführen, solange eine bestimmte Bedingung erfüllt ist. Sie sind ideal für wiederholte Aufgaben wie Zählen, Iteration über Sammlungen oder Verarbeitung von Daten.
Typen von Schleifen

Python kennt zwei Arten von Schleifen:

- for-Schleife: Wird verwendet, um über eine vordefinierte Anzahl von Schritten zu iterieren (z. B. über Elemente einer Liste).
 - while-Schleife: Wird solange ausgeführt, bis eine Bedingung nicht mehr erfüllt ist.


```python
fruits = ["Apfel", "Banane", "Kirsche"]
for fruit in fruits:
    print(fruit)
```
```python
# Zählt von 0 bis 4 (exklusiv 5)
for i in range(5):
    print(i)  # Ausgabe: 0,1,2,3,4
```
```python
count = 0
while count < 3:
    print(count)  # Ausgabe: 0,1,2
    count += 1    # Zähler inkrementieren
```
```python
for i in range(10):
    if i == 5:
        break  # Beendet die Schleife, wenn i=5
    print(i)   # Ausgabe: 0,1,2,3,4
```



Häufige Anwendungen

     Datenumgehung: Durchlaufen von Listen, Tupeln oder Dictionaries.
     Zählen: Zählen von Zahlen oder Inkrementieren eines Counters.
     Benutzerinteraktion: Wiederholtes Eingeben von Daten, bis eine Bedingung erfüllt ist.


## Funktionen
Funktionen helfen, Code zu modularisieren und wiederverwenden. Sie gruppieren Aufgaben in benannten Blöcken.
```python
def name_der_funktion(parameter1, parameter2):
    """Dokumentations-String (optional)."""
    # Code-Block
    return ergebnis  # Optional, gibt einen Wert zurück
```
```python
def begruessen(name, gruss="Hallo"):
    print(f"{gruss}, {name}!")

begruessen("Max")         # Ausgabe: Hallo, Max!
begruessen("Lena", "Servus")  # Ausgabe: Servus, Lena!
```

Wichtige Konzepte

1. Parameter:
    - Positional-Parameter: Werte an Positionen (z.B. name).
    - Standardwerte: Parameter mit voreingestellten Werten (z.B. gruss="Hallo").
    - Variadic Parameter: *args (beliebig viele Positional-Argumente), **kwargs (beliebig viele Keyword-Argumente).
2.  return:
    - Gibt einen Wert zurück (kann auch None sein).

## Dateien
Dateien ermöglichen das Speichern und Laden von Daten. Python bietet einfache Möglichkeiten, mit Dateien zu arbeiten.
```python
with open("datei.txt", "r") as datei:
    inhalt = datei.read()
    print(inhalt)
# Automatisch geschlossen nach dem Block!
```



Bedeutung | Befehl 
-------- |--------
 Schleifenkopf: while-Schleife  | `while x<10:`
 Schleifenkopf: for-Schleife  | `for i in range(10):`
 Funktionskopf mit 3 Parametern | `def my_function(a,b=1,c="wert):`
 Datei öffnen | `with open(path2file, 'r') as file:`
